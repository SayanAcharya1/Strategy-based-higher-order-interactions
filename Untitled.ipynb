## Code: fixed higher-order coupling strength $\lambda_2 = 5$, varying pairwise coupling strength $\lambda_1$

Coevolutionary pairwise + higher-order Kuramoto model on an ER network ($\langle k\rangle = 12$) and a 3-uniform hypergraph ($\langle k_h\rangle = 6$).

$$
\dot{\theta}_i = \omega_i + \lambda_1\, s^P_i \sum_{j} \sin(\theta_j - \theta_i) + \lambda_2\, s^H_i \sum_{(j,k)} \sin(2\theta_j - \theta_k - \theta_i)
$$

We scan $\lambda_1$ at fixed $\lambda_2 = 5$ and measure $\langle r_g \rangle$, $\langle s_P \rangle$, $\langle s_H \rangle$.

```julia
using Random, Statistics, DelimitedFiles, Printf
using CairoMakie
```

```julia
Base.@kwdef struct Params
    N::Int              = 1000
    T::Float64          = 1000.0
    dt::Float64         = 0.01
    trans_frac::Float64 = 0.8      # measure over the final (1 - trans_frac) of the run
    update_every::Int   = 1        # RK4 steps between payoff/strategy updates
    β::Float64          = 1.0      # Fermi intensity
    α::Float64          = 10.0^(-3.5)
    k_pair::Float64     = 12.0
    k_hyper::Float64    = 6.0
    λ2::Float64         = 5.0
    n_real::Int         = 50       # = number of fixed worlds
    theta0_scale::Float64 = 1e-3   # 1.0 => θ(0) uniform in [-π,π]; 1e-3 => near-synchronised start
end

nsteps(p::Params)           = Int(round(p.T / p.dt))
transient_steps(p::Params)  = Int(round(p.trans_frac * p.T / p.dt))

# ---- QUICK TEST (minutes) ----
P = Params(N = 100, T = 100.0, n_real = 3)

# ---- FULL RUN (very slow in serial) ----
# P = Params()   # N=1000, T=1000, dt=0.01, n_real=50, λ2=5, α=10^-3.5
```

```julia
struct RealizationData
    omega::Vector{Float64}
    neighbors_list::Vector{Vector{Int}}
    hyperedges::Vector{NTuple{3,Int}}
    hyper_per_node::Vector{Vector{Int}}
    hgroups::Vector{Vector{Tuple{Int,Int}}}
end

random_theta0(N, rng, scale) = (2π .* rand(rng, N) .- π)
random_uniform_omega(N, rng) = 2π .* rand(rng, N) .- π

function balanced_binary_strategy(N, rng)
    n1  = N ÷ 2
    arr = vcat(zeros(Int, N - n1), ones(Int, n1))
    shuffle!(rng, arr)
    return arr
end

function build_er_neighbors_list(N, k_avg, rng)
    p_edge = k_avg / (N - 1)
    nl = [Int[] for _ in 1:N]
    @inbounds for i in 1:N-1, j in i+1:N
        if rand(rng) < p_edge
            push!(nl[i], j); push!(nl[j], i)
        end
    end
    return nl
end

function build_fast_unique_hypergraph(N, k_hyper_avg, rng)
    M = Int(round(N * k_hyper_avg / 3))
    edge_set   = Set{NTuple{3,Int}}()
    hyperedges = NTuple{3,Int}[]
    while length(hyperedges) < M
        a = rand(rng, 1:N)
        b = rand(rng, 1:N); while b == a;            b = rand(rng, 1:N); end
        c = rand(rng, 1:N); while c == a || c == b;  c = rand(rng, 1:N); end
        x1 = min(a, b, c); x3 = max(a, b, c); x2 = a + b + c - x1 - x3
        edge = (x1, x2, x3)
        if !(edge in edge_set)
            push!(edge_set, edge); push!(hyperedges, edge)
        end
    end
    return hyperedges
end

function build_hyper_per_node(N, hyperedges)
    h = [Int[] for _ in 1:N]
    for (e, (a, b, c)) in enumerate(hyperedges)
        push!(h[a], e); push!(h[b], e); push!(h[c], e)
    end
    return h
end

function build_hyper_groups(N, hyperedges)
    g = [Tuple{Int,Int}[] for _ in 1:N]
    for (a, b, c) in hyperedges
        push!(g[a], (b, c)); push!(g[b], (a, c)); push!(g[c], (a, b))
    end
    return g
end

function prepare_fixed_worlds(p::Params; omega_seed_base = 100000, graph_seed_base = 200000)
    bank = Vector{RealizationData}(undef, p.n_real)
    for w in 1:p.n_real
        omega = random_uniform_omega(p.N, MersenneTwister(omega_seed_base + w))
        rng_g = MersenneTwister(graph_seed_base + w)
        nl    = build_er_neighbors_list(p.N, p.k_pair, rng_g)
        he    = build_fast_unique_hypergraph(p.N, p.k_hyper, rng_g)
        bank[w] = RealizationData(omega, nl, he,
                                  build_hyper_per_node(p.N, he),
                                  build_hyper_groups(p.N, he))
    end
    kp = mean(mean(length.(b.neighbors_list)) for b in bank)
    println("Prepared $(p.n_real) worlds | mean pair degree = ", round(kp, digits = 3),
            " | hyperedges/world = ", length(bank[1].hyperedges))
    return bank
end
```

```julia
function update_complex_phases!(z, θ)
    @inbounds for i in eachindex(θ)
        z[i] = cis(θ[i])
    end
end

function local_order_pairwise!(rl, z, nl)
    fill!(rl, 0.0)
    @inbounds for i in eachindex(z)
        neigh = nl[i]
        isempty(neigh) && continue
        s = 0.0
        for j in neigh
            s += abs((z[i] + z[j]) / 2)
        end
        rl[i] = s / length(neigh)
    end
end

function local_order_hyper!(rl, Rvals, z, hyperedges, hpn)
    fill!(rl, 0.0)
    @inbounds for (e, (i, j, k)) in enumerate(hyperedges)
        Rvals[e] = abs((z[i] + z[j] + z[k]) / 3)
    end
    @inbounds for i in eachindex(rl)
        idxs = hpn[i]
        isempty(idxs) && continue
        s = 0.0
        for e in idxs
            s += Rvals[e]
        end
        rl[i] = s / length(idxs)
    end
end

global_order(z) = abs(sum(z) / length(z))

struct RK4Buffers
    k1::Vector{Float64}; k2::Vector{Float64}; k3::Vector{Float64}; k4::Vector{Float64}
    θstage::Vector{Float64}
    pair_w::Vector{Float64}; ho_sums_w::Vector{Float64}; ho_forces_w::Vector{Float64}
end
make_rk4_buffers(N) = RK4Buffers((zeros(N) for _ in 1:8)...)

function kuramoto_rhs!(dθ, pair_forces, ho_sums, ho_forces, θ, sP, sH, ω, λ1, λ2, nl, hyperedges)
    fill!(ho_sums, 0.0)
    @inbounds for i in eachindex(θ)
        θi = θ[i]; ps = 0.0
        for j in nl[i]
            ps += sin(θ[j] - θi)
        end
        pair_forces[i] = sP[i] * λ1 * ps
    end
    @inbounds for (a, b, c) in hyperedges
        θa, θb, θc = θ[a], θ[b], θ[c]
        ho_sums[a] += sin(2θb - θc - θa) + sin(2θc - θb - θa)
        ho_sums[b] += sin(2θa - θc - θb) + sin(2θc - θa - θb)
        ho_sums[c] += sin(2θa - θb - θc) + sin(2θb - θa - θc)
    end
    @inbounds for i in eachindex(θ)
        ho_forces[i] = sH[i] * λ2 * ho_sums[i]
        dθ[i] = ω[i] + pair_forces[i] + ho_forces[i]
    end
end

function RK4_step!(θ, b::RK4Buffers, sP, sH, ω, λ1, λ2, nl, he, dt)
    rhs!(out, th) = kuramoto_rhs!(out, b.pair_w, b.ho_sums_w, b.ho_forces_w,
                                  th, sP, sH, ω, λ1, λ2, nl, he)
    rhs!(b.k1, θ)
    @. b.θstage = θ + 0.5dt * b.k1;  rhs!(b.k2, b.θstage)
    @. b.θstage = θ + 0.5dt * b.k2;  rhs!(b.k3, b.θstage)
    @. b.θstage = θ + dt * b.k3;     rhs!(b.k4, b.θstage)
    @. θ += (dt / 6) * (b.k1 + 2b.k2 + 2b.k3 + b.k4)
end
```

```julia
fermi(Δ, β) = 1.0 / (1.0 + exp(-β * Δ))

function simulate_one_realization(p::Params, w::RealizationData, λ1::Float64, seed::Int)
    N, dt, β, α, λ2 = p.N, p.dt, p.β, p.α, p.λ2
    ω, nl, he, hpn, hgroups = w.omega, w.neighbors_list, w.hyperedges, w.hyper_per_node, w.hgroups
    steps, tsteps = nsteps(p), transient_steps(p)

    rng = MersenneTwister(seed)
    θ  = random_theta0(N, rng, p.theta0_scale)
    sP = balanced_binary_strategy(N, rng)
    sH = balanced_binary_strategy(N, rng)
    sP_new = similar(sP); sH_new = similar(sH)

    rkbuf = make_rk4_buffers(N)
    z       = Vector{ComplexF64}(undef, N)
    rl_pair = zeros(N); rl_hyp = zeros(N); Rvals = zeros(length(he))

    dθ_now = zeros(N); dθ_prev = zeros(N)
    pf = zeros(N); hs = zeros(N); hf = zeros(N)
    c = zeros(N); c_pair = zeros(N); c_ho = zeros(N)
    ΠP = zeros(N); ΠH = zeros(N)

    # reference velocity so the first cost isn't |v - 0|
    kuramoto_rhs!(dθ_prev, pf, hs, hf, θ, sP, sH, ω, λ1, λ2, nl, he)

    rg_s = rg2_s = rp_s = rh_s = 0.0
    sp_s = sp2_s = sh_s = sh2_s = 0.0
    c_s = cp_s = ch_s = joint_s = glob_prod_s = 0.0
    saved = 0

    for step in 1:steps
        RK4_step!(θ, rkbuf, sP, sH, ω, λ1, λ2, nl, he, dt)

        if step % p.update_every == 0
            update_complex_phases!(z, θ)
            local_order_pairwise!(rl_pair, z, nl)
            local_order_hyper!(rl_hyp, Rvals, z, he, hpn)
            rg = global_order(z)

            kuramoto_rhs!(dθ_now, pf, hs, hf, θ, sP, sH, ω, λ1, λ2, nl, he)

            @inbounds for i in 1:N
                c[i]  = abs(dθ_now[i] - dθ_prev[i])
                tot   = abs(pf[i]) + abs(hf[i]) + 1e-12
                c_pair[i] = c[i] * abs(pf[i]) / tot
                c_ho[i]   = c[i] * abs(hf[i]) / tot
            end
            copyto!(dθ_prev, dθ_now)

            @inbounds for i in 1:N
                ΠP[i] = rl_pair[i] - α * c_pair[i] / (2π)
                ΠH[i] = rl_hyp[i]  - α * c_ho[i]   / (2π)
            end

            # ---- measurements (before strategies are overwritten) ----
            if step > tsteps
                sp_i = mean(sP); sh_i = mean(sH)
                rg_s  += rg;  rg2_s += rg^2
                rp_s  += mean(rl_pair); rh_s += mean(rl_hyp)
                sp_s  += sp_i; sp2_s += sp_i^2
                sh_s  += sh_i; sh2_s += sh_i^2
                c_s   += mean(c); cp_s += mean(c_pair); ch_s += mean(c_ho)
                joint_s     += sum(sP[i] * sH[i] for i in 1:N) / N
                glob_prod_s += sp_i * sh_i
                saved += 1
            end

            # ---- strategy update (synchronous) ----
            copyto!(sP_new, sP); copyto!(sH_new, sH)

            @inbounds for i in 1:N
                pn = nl[i]
                isempty(pn) && continue
                j = rand(rng, pn)
                if rand(rng) < fermi(ΠP[j] - ΠP[i], β)
                    sP_new[i] = sP[j]
                end
            end

            @inbounds for i in 1:N
                g = hgroups[i]
                isempty(g) && continue
                j, k = g[rand(rng, 1:length(g))]
                Πgroup = (ΠH[j] + ΠH[k]) / 2
                if rand(rng) < fermi(Πgroup - ΠH[i], β)
                    sH_new[i] = sH[j] == sH[k] ? sH[j] : (rand(rng, Bool) ? 1 : 0)
                end
            end

            sP, sP_new = sP_new, sP
            sH, sH_new = sH_new, sH
        end
    end

    saved == 0 && error("No long-time data saved. Check T, trans_frac, update_every.")

    rg_m = rg_s / saved; sp_m = sp_s / saved; sh_m = sh_s / saved
    std_rg = sqrt(max(rg2_s / saved - rg_m^2, 0.0))
    std_sp = sqrt(max(sp2_s / saved - sp_m^2, 0.0))
    std_sh = sqrt(max(sh2_s / saved - sh_m^2, 0.0))

    # node-level correlation
    sdp = sqrt(max(sp_m * (1 - sp_m), 0.0)); sdh = sqrt(max(sh_m * (1 - sh_m), 0.0))
    corr_local = (sdp < 1e-12 || sdh < 1e-12) ? 0.0 :
                 clamp((joint_s / saved - sp_m * sh_m) / (sdp * sdh), -1.0, 1.0)

    # temporal correlation of global fractions
    corr_global = (std_sp < 1e-12 || std_sh < 1e-12) ? 0.0 :
                  clamp((glob_prod_s / saved - sp_m * sh_m) / (std_sp * std_sh), -1.0, 1.0)

    return (rp = rp_s / saved, rh = rh_s / saved, rg = rg_m, sp = sp_m, sh = sh_m,
            std_rg = std_rg, std_sp = std_sp, std_sh = std_sh,
            chi_rg = N * std_rg^2, chi_sp = N * std_sp^2, chi_sh = N * std_sh^2,
            mean_c = c_s / saved, mean_cp = cp_s / saved, mean_ch = ch_s / saved,
            corr_local = corr_local, corr_global = corr_global)
end
```

```julia
function make_lambda_grid(; n1 = 500, n2 = 50, n3 = 500, n4 = 100)
    seg1 = 10.0 .^ (-2.0 .+ 2.5  .* range(0, 1, length = n1) .^ 0.6)
    seg2 = 10.0 .^ range(0.5, 1.0, length = n2)
    seg3 = 10.0 .^ (1.0  .+ 1.01 .* range(0, 1, length = n3) .^ 0.53)
    seg4 = 10.0 .^ range(2.01, 2.5, length = n4)
    all  = sort(vcat(seg1, seg2, seg3, seg4))
    keep = [1; (findall(i -> log10(all[i]) - log10(all[i-1]) > 1e-9, 2:length(all)) .+ 1)]
    return all[keep]
end

# quick test grid (~20 points, same shape as the full grid)
λ_grid = make_lambda_grid(n1 = 8, n2 = 3, n3 = 8, n4 = 4)

# full grid:
# λ_grid = make_lambda_grid()

println("Number of λ1 points = ", length(λ_grid))
```

```julia
function run_scan(p::Params, λ_grid; shuffle_seed = 700000)
    bank  = prepare_fixed_worlds(p)
    order = randperm(MersenneTwister(shuffle_seed), p.n_real)   # no world repeats within a λ1

    keys_ = (:rp, :rh, :rg, :sp, :sh, :std_rg, :std_sp, :std_sh,
             :chi_rg, :chi_sp, :chi_sh, :mean_c, :mean_cp, :mean_ch,
             :corr_local, :corr_global)
    out = zeros(length(λ_grid), 3 + length(keys_))

    t0 = time()
    for (idx, λ1) in enumerate(λ_grid)
        acc = zeros(length(keys_))
        for r in 1:p.n_real
            seed = 999_983 * idx + 1009 * r           # unique across the whole scan
            res  = simulate_one_realization(p, bank[order[r]], λ1, seed)
            acc .+= collect(getfield(res, k) for k in keys_)
        end
        out[idx, 1:3]   = [λ1, p.λ2, p.α]
        out[idx, 4:end] = acc ./ p.n_real
        @printf("[%3d/%3d] λ1 = %9.4f | rg = %.4f | sP = %.3f | sH = %.3f | %.0fs elapsed\n",
                idx, length(λ_grid), λ1, out[idx, 6], out[idx, 7], out[idx, 8], time() - t0)
    end
    return out
end

results = run_scan(P, λ_grid)
```
